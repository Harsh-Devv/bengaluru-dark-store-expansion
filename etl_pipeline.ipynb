```python
import pandas as pd
```

```python
pop = pd.read_csv("data/raw/bengaluru_wards_2011.csv")
rent = pd.read_csv("data/raw/commercial_rent.csv")
comp = pd.read_csv("data/raw/competitor_density.csv")
```

```python
print(pop.head(2))
```

```python
pop = pop.drop(pop.columns[4:12], axis=1)
```

```python
print(pop.head(2))
```

```python
zone_ranges = {
    "Yelahanka" : range(1, 14),
    "Bangalore North" : range(14,35),
    "Dasarahalli" : range(35, 45),
    "Bangalore West" : list(range(45,49)) + list(range(64, 81)) + list(range(94, 107)) + list(range(107,109)) + list(range(120,142)),
    "Bangalore East" :  list(range(49,64)) + list(range(81, 94)) + list(range(109, 120)),
    "Bangalore South" : range(142, 186),
    "Bommanahalli" : range(186, 194),
    "Mahadevapura" : list(range(194, 199)) + list(range(82,86))
}

ward_to_zone = {
    ward : zone for zone, wards in zone_ranges.items() for ward in wards
}
```

```python
pop['Zone'] = pop['Ward Num'].map(ward_to_zone)
```

```python
print(pop.head(5))
```

```python
pop['Density Tier'] = pd.qcut(pop['Population'], q=3, labels=['Low Density', 'Medium Density', 'High Density'])
```

```python
print(pop.sample(5))
```
